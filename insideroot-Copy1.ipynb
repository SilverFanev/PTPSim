```python
import re
from collections import defaultdict
```

```python
def extract_trees_and_branches(cpp_files):
    # Dictionnaire : nom de la variable C++ -> nom du TTree ROOT
    # ex: 'metaTree' -> 'meta'
    tree_map = {}
    
    # Dictionnaire : nom du TTree ROOT -> liste des branches
    # ex: 'meta' -> ['"runId"', '"runType"', ...]
    tree_branches = defaultdict(list)

    # Regex 1 : Création du TTree (ex: metaTree = new TTree("meta", "meta");)
    tree_pattern = re.compile(r'\b([a-zA-Z_]\w*)\s*=\s*new\s+TTree\s*\(\s*"([^"]+)"')
    
    # Regex 2 : Utilisation de MakeBranch (ex: MakeBranch(metaTree, "runId", &runId);)
    make_branch_pattern = re.compile(r'MakeBranch\s*\(\s*([a-zA-Z_]\w*)\s*,\s*([^,]+)\s*,')
    
    # Regex 3 : Utilisation de ->Branch (ex: this->metaTree->Branch("geo_index", &geo_index);)
    # Le (?:this->)? permet d'ignorer "this->" s'il est présent.
    arrow_branch_pattern = re.compile(r'(?:this->)?([a-zA-Z_]\w*)->Branch\s*\(\s*([^,]+)\s*,')

    # On lit le contenu de tous les fichiers et on les concatène.
    # Cela permet de lier la création (dans ito.cc) et l'utilisation (dans NtupleProc.cc).
    full_code = ""
    for file_path in cpp_files:
        try:
            with open(file_path, 'r', encoding='utf-8') as f:
                full_code += f.read() + "\n"
        except FileNotFoundError:
            print(f"Attention : Le fichier {file_path} est introuvable.")

    # 1. Identifier tous les TTrees
    for match in tree_pattern.finditer(full_code):
        cpp_var = match.group(1)
        root_name = match.group(2)
        tree_map[cpp_var] = root_name

    # Fonction interne pour ajouter une branche au bon arbre
    def add_branch_to_tree(cpp_var, branch_expr):
        branch_expr = branch_expr.strip()
        # Si l'arbre a été trouvé à l'étape 1, on utilise son vrai nom,
        # sinon on garde le nom de la variable (par sécurité).
        root_name = tree_map.get(cpp_var, cpp_var)
        if branch_expr not in tree_branches[root_name]:
            tree_branches[root_name].append(branch_expr)

    # 2. Chercher les MakeBranch(...)
    for match in make_branch_pattern.finditer(full_code):
        cpp_var = match.group(1)
        branch_expr = match.group(2)
        add_branch_to_tree(cpp_var, branch_expr)

    # 3. Chercher les ->Branch(...)
    for match in arrow_branch_pattern.finditer(full_code):
        cpp_var = match.group(1)
        branch_expr = match.group(2)
        add_branch_to_tree(cpp_var, branch_expr)

    return tree_map, tree_branches
```

```python
def print_report(tree_map, tree_branches):
    print("="*60)
    print(" 🌳 ANALYSE DES TTrees ROOT ET DE LEURS BRANCHES 🌳")
    print("="*60)
    
    print("\n[TTREES IDENTIFIÉS]")
    for cpp_var, root_name in tree_map.items():
        print(f" - Variable C++ : '{cpp_var}'  ==>  Nom ROOT : '{root_name}'")

    print("\n" + "="*60)
    
    for tree_name, branches in tree_branches.items():
        print(f"\n📂 ARBRE ROOT : {tree_name}")
        print("-" * (16 + len(tree_name)))
        
        static_branches = []
        dynamic_branches = []
        
        # On sépare les branches statiques ("nom") des dynamiques ("nom" + var)
        for b in branches:
            if '+' in b or not (b.startswith('"') and b.endswith('"')):
                dynamic_branches.append(b)
            else:
                static_branches.append(b.strip('"'))
                
        print(f"  📌 Branches Fixes (Statiques) [{len(static_branches)}] :")
        if static_branches:
            # Affichage sur plusieurs colonnes pour gagner de la place
            for i in range(0, len(static_branches), 3):
                row = static_branches[i:i+3]
                print("      " + ", ".join([f'"{name}"' for name in row]))
        else:
            print("      Aucune")
            
        print(f"\n  ⚙️  Branches Dynamiques (Générées via boucle/code) [{len(dynamic_branches)}] :")
        if dynamic_branches:
            for b in dynamic_branches:
                print(f"      -> {b}")
        else:
            print("      Aucune")
    
    print("\n" + "="*60)
```

```python
def extraire_branches_cpp(fichiers_cpp):
    """
    Extrait les noms de branches statiques et dynamiques depuis les fichiers C++.
    """
    make_branch_pattern = re.compile(r'MakeBranch\s*\(\s*[a-zA-Z_]\w*\s*,\s*([^,]+)\s*,')
    arrow_branch_pattern = re.compile(r'(?:this->)?[a-zA-Z_]\w*->Branch\s*\(\s*([^,]+)\s*,')
    
    code_complet = ""
    for fichier in fichiers_cpp:
        try:
            with open(fichier, 'r', encoding='utf-8') as f:
                code_complet += f.read() + "\n"
        except FileNotFoundError:
            print(f"Attention : Le fichier {fichier} est introuvable.")

    toutes_les_branches = []
    toutes_les_branches.extend(make_branch_pattern.findall(code_complet))
    toutes_les_branches.extend(arrow_branch_pattern.findall(code_complet))
    
    branches_statiques = set()
    branches_dynamiques = set()
    
    for b in toutes_les_branches:
        b = b.strip()
        # Si c'est une chaîne statique pure (ex: "mcpdg")
        if b.startswith('"') and b.endswith('"') and '+' not in b:
            # On enlève les guillemets
            branches_statiques.add(b.strip('"'))
        else:
            branches_dynamiques.add(b)
            
    return branches_statiques, branches_dynamiques


def comparer_dataframe_et_cpp(df, branches_statiques, branches_dynamiques):
    """
    Compare les colonnes du DataFrame avec les branches extraites du C++.
    """
    # 1. Récupération des noms de colonnes du DataFrame
    # Si tes noms sont bien dans l'en-tête (standard pandas) :
    colonnes_df = set(df.columns)
    
    # Si tes noms étaient littéralement sur la première ligne (index 0) du DataFrame :
    # colonnes_df = set(df.iloc[0].values)
    
    print("="*60)
    print(" 🔍 RAPPORT DE COMPARAISON : DATAFRAME VS C++")
    print("="*60)

    # --- COMPARAISON DES BRANCHES STATIQUES ---
    manquantes_statiques = branches_statiques - colonnes_df
    presentes_statiques = branches_statiques.intersection(colonnes_df)
    
    print(f"\n[BRANCHES FIXES] : {len(manquantes_statiques)} absentes sur {len(branches_statiques)} attendues")
    if manquantes_statiques:
        print("Les variables suivantes sont dans le C++ mais PAS dans votre DataFrame :")
        for var in sorted(manquantes_statiques):
            print(f"  ❌ {var}")
    else:
        print("  ✅ Toutes les branches fixes du C++ sont présentes dans le DataFrame !")


    # --- COMPARAISON DES BRANCHES DYNAMIQUES ---
    print(f"\n[BRANCHES DYNAMIQUES] : Analyse des {len(branches_dynamiques)} expressions générées par le code")
    
    # On va extraire les préfixes/suffixes textes des expressions dynamiques 
    # Ex: '"fit_pmtid_" + fitter_name' -> on extrait 'fit_pmtid_'
    fragments_introuvables = []
    
    for expr in branches_dynamiques:
        # On cherche tous les bouts de texte entre guillemets
        fragments_texte = re.findall(r'"([^"]+)"', expr)
        
        for fragment in fragments_texte:
            # On ignore les tirets bas isolés (trop génériques pour chercher)
            if fragment == "_":
                continue
                
            # Vérifie si AUCUNE colonne du DF ne contient ce fragment
            if not any(fragment in col for col in colonnes_df):
                fragments_introuvables.append((fragment, expr))
                
    if fragments_introuvables:
        print("Les fragments de noms suivants (issus de boucles C++) sont INTROUVABLES dans le DataFrame :")
        for fragment, expr in sorted(set(fragments_introuvables)):
            print(f"  ❌ Fragment '{fragment}' (issu de l'expression C++ : {expr})")
    else:
        print("  ✅ Le DataFrame semble contenir des colonnes pour toutes les branches dynamiques !")

    print("\n" + "="*60)
```
