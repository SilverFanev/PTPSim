```python
import uproot
import pandas as pd


important_variables = ['mcnhits', 'mcPEWavelength','mcpecount']
```

```python
#10_000 photons and 100 events (One million photons)
with uproot.open("/home/silver/PTPSim/results/output.ntuple_10000photons_100event.root") as file:
    print(file.keys())
    tree =  file["output"]
    df_5 = tree.arrays(library="pd")

sum_mcnhist_5 = 0
sum_nb_wvlght_5 = 0
sum_mcpecount_5 = 0
sum_all_PE_treated_5 = 0
for j in range(len(df_5[important_variables[0]])):
    sum_mcnhist_5 = sum_mcnhist_5 + df_5[important_variables[0]][j]
    sum_nb_wvlght_5 = sum_nb_wvlght_5 +len(df_5[important_variables[1]][j])
    sum_mcpecount_5 = sum_mcpecount_5 + df_5[important_variables[2]][j]
    sum_all_PE_treated_5 = sum_all_PE_treated_5 + df_5['mcparticlecount'][j]
print(f'There was {sum_mcnhist_5} total values in mcnhits and there was {sum_nb_wvlght_5} values in the wavelentgh, and has {sum_mcpecount_5} mcpacount  over {sum_all_PE_treated_5} in total')
print(f'So the efficiency may be {(sum_mcpecount_5*100)/sum_all_PE_treated_5}%')
```

```python
data_wvlght["Intervalle_Num"] = (
    data_wvlght['Wavelenght'] // 0.000005
) * 5
```

```python
for_plot = pd.DataFrame(
    {"Wavelenght": sorted(data_wvlght["Intervalle_Num"].unique())}
)
```

```python
for_plot = pd.DataFrame(
    {"Wavelenght": sorted(data_wvlght["Intervalle_Num"].unique())}
)
```

```python
len(for_plot['Wavelenght'])
```

```python
for_plot = (
    data_wvlght["Intervalle_Num"]
    .value_counts()
    .reset_index(name="Compte")
)
```

```python
for_plot.head()
```

```python
import matplotlib.pyplot as plt

# 1. Configurer la taille du graphique
plt.figure(figsize=(12, 6))

# 2. Dessiner le diagramme en barres
# width=0.0000045 permet de laisser un tout petit espace noir entre les barres pour le style
plt.bar(
    for_plot["Intervalle_Num"],
    for_plot["Compte"],
    width=5,
    color="white",
    edgecolor="skyblue",
    align="edge",
)

# 4. Personnaliser les titres et les axes
plt.title(
    "Distribution of Wavelengths",
    fontsize=14,
    fontweight="bold",
)
plt.xlabel("Wavelength Interval (nm)", fontsize=12)
plt.ylabel("Nombre of time an interval showing", fontsize=12)

# Évite que les étiquettes du bas ou des côtés soient coupées
plt.tight_layout()

# 6. Afficher le graphique
plt.show()
```
