```python
with uproot.open("/home/silver/PTPSim/output.root") as file:
    # 1. Parcourir tous les objets du fichier ROOT
    for key, classname in file.classnames().items():
        
        # 2. Vérifier si l'objet est un TTree
        if classname == "TTree":
            # Nettoyer le nom de la clé (enlever le numéro de cycle comme ';1')
            tree_name = key.split(";")[0]
            
            print(f"\n===== Branches de l'arbre : {tree_name} =====")
            
            # 3. Récupérer l'arbre et lister ses branches
            tree = file[key]
            branches = tree.keys()
            if branches:
                for branch in branches:
                    print(f"  -> {branch}")
            else:
                print("  (Aucune branche trouvée)")
            tree = file["runT"]
            df = tree.arrays(library="pd")
```
